```python
from random import sample
from random import randint
from random import randrange
from matplotlib.pyplot import bar, hist, grid, show, legend
```

```python
def MontyHall(schimba_usa, nr_sim, afisare):
    contor = 0
    for _ in range(nr_sim):
        asezare_initiala = ['c', 'c', 'c']
        masina, capra1, capra2 = sample([0, 1, 2], 3)
        asezare_initiala[masina] = 'm'
        alegere_initiala = randint(0, 2)
        if alegere_initiala == masina:
            alegere_prezentator = sample([capra1, capra2], 1).pop()
        elif alegere_initiala == capra1:
            alegere_prezentator = capra2
        else:
            alegere_prezentator = capra1

        if schimba_usa:
            alegere_finala = ({0, 1, 2} - {alegere_initiala, alegere_prezentator}).pop()
        else:
            alegere_finala = alegere_initiala

        contor += alegere_finala == masina

    return contor / nr_sim
        
```

```python
print(MontyHall(True, 1000, False))
```

```python
data = [randrange(1,7) for _ in range(500)]
bin_edges = [k+0.5 for k in range(7)]

hist(data, bins=bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',
    alpha = 0.5, label = 'frecvente relative')

distribution = dict([(i,1/6) for i in range(1,7)])
bar(list(distribution.keys()), list(distribution.values()), width = 0.85, color = 'red', edgecolor = 'black',
        alpha= 0.6, label = 'probabilitati')

legend(loc = 'lower left')
grid()
show()
```

```python
from scipy.stats import binom
# help('scipy.stats.binom')
```

```python
n = 5
p = 0.6
x = binom.rvs(n, p, size=1000)
bin_edges = [k+0.5 for k in range(6)]
hist(x, bins=bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',
    alpha = 0.5, label = 'frecvente relative')

distribution = dict([(i,binom.pmf(i, 5, 6/10)) for i in range(1,6)])
bar(list(distribution.keys()), list(distribution.values()), width = 0.85, color = 'red', edgecolor = 'black',
        alpha= 0.6, label = 'probabilitati')

legend(loc = 'lower left')
grid()
show()
```
