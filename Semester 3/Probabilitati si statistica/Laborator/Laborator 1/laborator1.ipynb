```python
from random import sample
from math import factorial, perm, comb
from itertools import permutations, combinations, combinations_with_replacement
```

```python
#1a
print(list(permutations('word')))
```

```python
#1b
print(factorial(len('word')))
```

```python
#1c
print(sample('word', 4))
```

```python
def aranjamente(text, k, numar_total=False, aleator=False):
    if numar_total:
        return factorial(len(text)) // factorial(k)
    elif aleator:
        return sample(text, k)
    return(list(permutations(text, k)))
```

```python
#2a
print(aranjamente('word', 2))
```

```python
print(aranjamente('word', 2, numar_total=True))
```

```python
print(aranjamente('word', 2, aleator=True))
```

```python
def combinari(text, k, numar_total=False, aleator=False):
    if numar_total:
        return comb(len(text), k)
    if aleator:
        return sample(text, k)
    return list(combinations(text, k))
```

```python
#2b
print(combinari('word', 3))
```

```python
print(combinari('word', 3, numar_total=True))
```

```python
print(combinari('word', 3, aleator=True))
```

```python
print(list(combinations_with_replacement('abbe', 3)))
```
