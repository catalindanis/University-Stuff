```python
from scipy.stats import uniform, expon
from matplotlib.pyplot import hist, show, grid, bar, plot, xticks
```

```python
def generator(x, p, n):
    rezultat = []
    for _ in range(n):
        y = uniform.rvs()
        sum = 0
        for i in range(len(p)):
            if sum <= y <= sum + p[i]:
                X = x[i]
            sum += p[i]
        rezultat.append(X)
    return rezultat
```

```python
grupe = ['A', 'B', '0', 'AB']
p = [0.4, 0.1, 0.46, 0.04]
```

```python
data = generator(grupe, p, 10)
hist(data, bins=4)
bar(grupe, p, color='green', width=0.3)
grid()
show()
```

```python
def generare(l, n):
    return expon.rvs(scale=1/l, size=n)
```

```python
data = generare(1/12, 1000)
hist(data, bins=12, density=True , range=(0, 61))
x = range(60)
plot(x,expon.pdf(x,loc=0,scale=12),'r-')
xticks(range(0,60,5))
grid()
show()
```
