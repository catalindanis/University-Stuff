```python
from scipy.stats import norm, uniform, expon
from matplotlib.pyplot import hist, grid, show, bar, xticks, plot
from numpy import linspace, mean, std
```

```python
def simulare(n, m, d):
    return norm.rvs(loc=m, scale=d, size=n)
```

```python
data = simulare(1000, 165, 10)
x = linspace(130, 210, 1000)
plot(x, norm.pdf(x, loc=165, scale=10))
hist(data, bins=16, range=(130,210), density=True)
xticks(range(130, 210, 5))
grid()
show()
```

```python
data = simulare(1000, 165, 10)
avg = mean(data)
dev = std(data)
prop = mean([160 <= k <= 170 for k in data])
```

```python
print(avg)
print(dev)
print(prop)
```

```python
print(norm.cdf(170, 165, 10) - norm.cdf(160, 165, 10))
```

```python
def printare():
    x = uniform.rvs()
    if x < 0.4:
        return expon.rvs(5)
    else:
        return uniform.rvs(4, 2)
```

```python
data = [printare() for _ in range(1000)]
print(mean(data))
print(std(data))
```

```python
data = [printare() for _ in range(1000)]
prob = mean([k < 5 for k in data])
print(prob)
```
