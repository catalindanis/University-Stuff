```python
from scipy.stats import norm, expon
from numpy import mean, std, linspace
from matplotlib.pyplot import show, hist, grid, legend, xticks, plot
from numpy import linspace
```

```python
m = 165
s = 10
data = norm.rvs(loc=m, scale=s, size=1000)
```

```python
x = linspace(130, 210, 1000)
plot(x,norm.pdf(x, loc = m, scale = s), '-r')
hist(data, bins=16, density=True, range=(130, 210))
xticks(range(130, 210, 5), fontsize=8)
grid()
show()
```

```python
norm.cdf(170, loc=m, scale=s) - norm.cdf(160, loc=m, scale=s)
```

```python
from scipy.stats import expon, uniform
from numpy import mean, std, multiply
```

```python
data1 = expon.rvs(loc=0, scale=5, size=1000)
data2 = uniform.rvs(loc=4, scale=2, size=1000)

def getRandomTime():
    Y = uniform.rvs()
    if Y < 0.4:
        T = expon.rvs(loc=0, scale=5)
    else:
        T = uniform.rvs(loc=4, scale=2)
    return T

data = [getRandomTime() for _ in range(1000)]

print(mean(data))
print(std(data))

prob = [d < 5 for d in data]
print(mean(prob))

print(expon.cdf(5, loc=0, scale=5) * 0.4 + uniform.cdf(5, loc=4, scale=2) * 0.6)
```
