```python
from scipy.stats import bernoulli
from matplotlib.pyplot import hist, show, xticks
```

```python
def deplasari(n, p):
    moves = bernoulli.rvs(p, size=n)
    current_pos = 0
    positions = [0]
    for move in moves:
        current_pos += 2 * move - 1
        positions.append(int(current_pos))
    return positions
```

```python
n = 1000
p = 0.5
data = deplasari(n, p)
bin_edges = [k + 0.5 for k in range(min(data) - 1, max(data) + 1)]
hist(data, bins=bin_edges)
show()
```

```python
def deplasari_cerc(n, p):
    moves = bernoulli.rvs(p, size=n)
    current_pos = 0
    positions = [0]
    for move in moves:
        # print("Current pos: ", current_pos, ", ", "Current move: ", move, ", ")
        current_pos = (current_pos + (2 * move - 1)) % n
        positions.append(int(current_pos))
    return positions
```

```python
data = deplasari_cerc(n, p)
bin_edges = [k + 0.5 for k in range(-1, n + 1)]
hist(data, bins = bin_edges)
show()
```

```python
from scipy.stats import geom, hypergeom
```

```python
def generare(size):
    M = 49
    n1 = 6
    n2 = 6
    prob = sum(hypergeom.pmf(k, M, n1, n2) for k in range(3, 7))
    result = geom.rvs(prob, size=size) - 1
    return result
```

```python
print(generare(2))
```

```python
def cerinta_ii():
    n = 1000
    aproximate = sum(k > 10 for k in generare(n))/ n
    prob = sum(hypergeom.pmf(k, 49, 6, 6) for k in range(3, 7))
    theoretical = 1 - geom.cdf(10, prob)
    return float(aproximate), float(theoretical)
```

```python
print(cerinta_ii())
```
