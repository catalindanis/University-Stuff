```python
from random import randint
from matplotlib.pyplot import plot, grid, show
```

```python
def simulare_aceeasi_zi(n, ns):
    count = 0
    for _ in range(ns):
        birthdays = [randint(1, 365) for _ in range(n)]
        if len(birthdays) != len(set(birthdays)):
            count += 1
    return count / ns
```

```python
simulare_aceeasi_zi(23, 1000)
```

```python
def probabilitate_aceeasi_zi(n):
    fav_cases = 365
    total_cases = 365
    result = 1
    for _ in range(n):
        result = result * (fav_cases / total_cases)
        fav_cases -= 1
    return 1 - result
```

```python
probabilitate_aceeasi_zi(30)
```

```python
n = 50
xs = range(2, 51)
y1 = [simulare_aceeasi_zi(x, 1000) for x in xs]
y2 = [probabilitate_aceeasi_zi(x) for x in xs]
plot(xs, y1, 'r')
plot(xs, y2, 'b')
show()
```
