```python
from random import randint
```

```python
#1a
def same_birthday_simulations(n, ns):
    count = 0
    for _ in range(ns):
        birthdays = [randint(1, 365) for _ in range(n)]
        if len(birthdays) != len(set(birthdays)):
            count += 1
    return count / ns

same_birthday_simulations(23, 1000)
```

```python
#1c
def same_birthday_probability(n):
    favorite_cases = 365
    total_cases = 365
    result = 1
    for _ in range(n):
        result = result * (favorite_cases / total_cases)
        favorite_cases -= 1

    return 1 - result

same_birthday_probability(30)
```

```python
from matplotlib.pyplot import plot, grid, title, show, axis
from random import random
from math import dist
```

```python
#1c
title('Plot test')
xs = range(2, 51)
ys = [same_birthday_probability(x) for x in xs]
ys1 = [same_birthday_simulations(x, 1000) for x in xs]
plot(xs,ys,'r')
plot(xs, ys, 'b')
grid()
show()
```

```python
#2
def puncte_patrat(N, cerinta):
    axis((0, 1, 0, 1))
    count = 0
    A, B, C, D, E = [0, 0], [1, 0], [1, 1], [0, 1], [0.5, 0.5]
    match cerinta:
        case 1:
            for _ in range(N):
                F = [random(), random()]
                if dist(F, E) <= 0.5:
                    count += 1
                    plot(F[0], F[1], 'ro')
                else:
                    plot(F[0], F[1], 'b*')
            show()
            return count / N
        case 2:
            for _ in range(N):
                F = [random(), random()]
                if dist(F, E) <= min(dist(F, A), dist(F, B), dist(F, C), dist(F, D)):
                    count += 1
                    plot(F[0], F[1], 'ro')
                else:
                    plot(F[0], F[1], 'b*')
            show()
            return count / N
        case 3:
            for _ in range(N):
                F = [random(), random()]
                nr_tr_obtuze = (dist(A, F) * dist(A, F) + dist(B, F) * dist(B, F) < 1) + \
                    (dist(B, F) * dist(B, F) + dist(C, F) * dist(C, F) < 1) + \
                    (dist(D, F) * dist(D, F) + dist(C, F) * dist(C, F) < 1) + \
                    (dist(D, F) * dist(D, F) + dist(A, F) * dist(A, F) < 1)

                if nr_tr_obtuze == 2:
                    count += 1
                    plot(F[0], F[1], 'ro')
            return count / N
            
    
puncte_patrat(1000, 3)
```
